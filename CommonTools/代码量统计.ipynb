```python
import os
from pathlib import Path

def count_code_lines(file_path):
    """统计单个C/C++文件的代码行数"""
    code_lines = 0       # 有效代码行
    blank_lines = 0      # 空行
    comment_lines = 0    # 注释行
    in_multi_comment = False  # 是否在多行注释/*...*/中

    try:
        # 读取文件，忽略编码错误兼容不同格式
        with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
            for line in f:
                stripped_line = line.strip()
                
                # 处理空行
                if not stripped_line:
                    blank_lines += 1
                    continue
                
                # 处理多行注释
                if in_multi_comment:
                    comment_lines += 1
                    if '*/' in stripped_line:
                        in_multi_comment = False
                        # 注释后有代码则算有效行
                        remaining = stripped_line.split('*/')[1].strip()
                        if remaining and not remaining.startswith('//'):
                            code_lines += 1
                    continue
                
                # 单行注释
                if stripped_line.startswith('//'):
                    comment_lines += 1
                    continue
                
                # 代码+单行注释混合行
                if '//' in stripped_line:
                    code_part = stripped_line.split('//')[0].strip()
                    if code_part:
                        code_lines += 1
                    comment_lines += 1
                    continue
                
                # 多行注释开始
                if '/*' in stripped_line:
                    comment_lines += 1
                    if '*/' in stripped_line:
                        # 注释前后有代码则算有效行
                        if stripped_line.split('/*')[0].strip():
                            code_lines += 1
                    else:
                        in_multi_comment = True
                    continue
                
                # 有效代码行
                code_lines += 1

    except Exception as e:
        print(f"⚠️ 读取文件 {file_path} 出错: {e}")
        return 0, 0, 0

    return code_lines, blank_lines, comment_lines

def main():
    # ==================== 只需修改这里的目录路径 ====================
    # 替换成你要统计的代码目录（绝对路径/相对路径都可以）
    TARGET_DIR = "E:\\Code\\codeGit\\ANewCode\\GameClient\\src"  # 示例：当前目录下的my_cpp_project文件夹
    # ==============================================================

    # 检查目录是否存在
    if not os.path.isdir(TARGET_DIR):
        print(f"❌ 错误：目录 {TARGET_DIR} 不存在！请检查路径是否正确")
        return

    # 初始化统计数据
    total_code = 0
    total_blank = 0
    total_comment = 0
    file_count = 0

    print(f"🔍 开始扫描目录：{TARGET_DIR}")
    print("----------------------------------------")

    # 递归扫描所有.c/.cpp文件
    for file_path in Path(TARGET_DIR).rglob('*'):
        if file_path.suffix in ['.c', '.cpp'] and file_path.is_file():
            file_count += 1
            code, blank, comment = count_code_lines(file_path)
            total_code += code
            total_blank += blank
            total_comment += comment
            # 可选：显示每个文件的统计（不想看可以注释掉这行）
            print(f"📄 {file_path} | 代码：{code} | 空行：{blank} | 注释：{comment}")

    # 输出汇总结果
    print("----------------------------------------")
    print("\n📊 统计汇总：")
    print(f"📝 共处理 {file_count} 个C/C++文件")
    print(f"💻 有效代码行：{total_code} 行")

    print(f"📄 空行：{total_blank} 行")
    print(f"📝 注释行：{total_comment} 行")
    print(f"🧮 总行数：{total_code + total_blank + total_comment} 行")

if __name__ == '__main__':
    main()
```
